```python
from moviepy import VideoClip
import numpy as np
import random

# --- Parameters ---
duration = 6.0                     # total video duration (seconds)
fix_duration = 1.5                 # fixation ON duration
gap_duration = 0.5                 # gap (no fixation)
fps = 30                           # frames per second
size = (960, 540)                  # video dimensions (width, height)
bg_color = (128, 128, 128)         # gray background
fix_colors = [
    (255, 0, 0),    # red
    (0, 255, 0),    # green
    (0, 0, 255),    # blue
    (255, 255, 0),  # yellow
    (255, 0, 255)   # magenta
]

# --- Fixation schedule ---
fix_times = [(0, 1.5), (2.0, 3.5), (4.0, 5.5)]  # ON periods
fix_sequence = [random.choice(fix_colors) for _ in fix_times]

# --- Frame drawing function ---
def make_frame(t):
    # start with gray background
    frame = np.ones((size[1], size[0], 3), dtype=np.uint8) * np.array(bg_color, dtype=np.uint8)
    
    # check if fixation should be visible
    for i, (start, end) in enumerate(fix_times):
        if start <= t < end:
            color = fix_sequence[i]
            cx, cy = size[0] // 2, size[1] // 2
            cross_len = 40
            cross_thick = 6
            
            # horizontal bar
            frame[cy - cross_thick//2 : cy + cross_thick//2, cx - cross_len//2 : cx + cross_len//2] = color
            # vertical bar
            frame[cy - cross_len//2 : cy + cross_len//2, cx - cross_thick//2 : cx + cross_thick//2] = color
            break
    return frame

# --- Create and write video ---
clip = VideoClip(make_frame, duration=duration)
clip.write_videofile("fixation_movie.mp4", fps=fps, codec="libx264", audio=False)
```
